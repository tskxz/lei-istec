## 1. Shared features and the key difference

A list literal uses square brackets. A tuple literal commonly uses parentheses. Both preserve the order of their items and can contain objects of different types.

| Feature | List | Tuple |
|---|---|---|
| Ordered sequence | Yes | Yes |
| Can hold arbitrary Python objects | Yes | Yes |
| Supports indexing and slicing | Yes | Yes |
| Can contain nested sequences | Yes | Yes |
| Supports in-place changes to its item positions | Yes | No |

A list is mutable. You can replace, insert, and remove items. A tuple is immutable: you cannot change its sequence of item references after construction.

### Ordered does not mean sorted

An ordered sequence retains a positional arrangement; Python does not automatically sort its items. A mutable list can later be rearranged. A tuple retains its sequence of references.

### List comprehensions

A list comprehension builds a list by evaluating an expression for selected items from an iterable. This example keeps the even numbers from 1 through 9.

```python
even_digits = [number for number in range(1, 10) if number % 2 == 0]
print(even_digits)
print([])
print(list())
```

### Tuple literals and the importance of commas

For a nonempty tuple, commas separate the items. Parentheses improve readability and can be needed by the surrounding syntax. A one-item tuple needs a trailing comma; parentheses alone merely group an expression.

```python
connection = ("localhost", "8080", 3, "database.db")
contact = "John Doe", "john@example.com", "55-555-5555"
print(connection)
print(contact)

one_item = (2,)
not_a_tuple = (2)
print(one_item, type(one_item).__name__)
print(not_a_tuple, type(not_a_tuple).__name__)
```

## List methods

List methods offer explicit ways to add and remove items. Tuples do not provide these mutating methods.

| Method | Effect | Return value |
|---|---|---|
| `append(obj)` | Add one object at the end | `None` |
| `extend(iterable)` | Add each item from an iterable | `None` |
| `insert(index, obj)` | Insert one object at a position | `None` |
| `remove(obj)` | Remove the first matching value | `None` |
| `pop(index=-1)` | Remove an item by position | The removed item |

Do not assume that every mutating method returns `None`: `pop()` is an important exception.

### `append()`: add one object

Appending a list adds that list as one nested item. It does not add each of the nested list’s elements separately.

```python
a = ["a", "b"]
result = a.append("c")
print(a)
print("Returned:", result)

a = ["a", "b"]
a.append(["c", "d", "e"])
print(a)
print("Number of outer items:", len(a))
```

### `extend()`: add the items from an iterable

`extend()` adds individual items at the end of the existing list. Augmented concatenation with `+=` can produce the same list contents.

```python
a = ["a", "b"]
a.extend(["c", "d", "e"])
print(a)

a = ["a", "b"]
a += ["c", "d", "e"]
print(a)
```

### `insert()`: add one object at a position

Inserting before the item currently at index 1 moves that item and later items to the right.

```python
a = ["a", "c"]
a.insert(1, "b")
print(a)
```

### `remove()`: remove the first matching value

The argument is the value to remove, not its position. A duplicated value loses only its first occurrence. A missing value raises `ValueError`.

```python
a = ["a", "b", "c", "d", "e"]
a.remove("b")
print(a)
a.remove("c")
print(a)

repeated = ["bark", "meow", "bark"]
repeated.remove("bark")
print(repeated)
try:
    a.remove("missing")
except ValueError as error:
    print(type(error).__name__ + ":", error)
```

### `pop()`: remove and return an item

Without an index, `pop()` removes the final item. Unlike `remove()`, its argument identifies a position and its return value is the removed object.

```python
a = ["a", "b", "c", "d", "e"]
removed = a.pop()
print("Removed:", removed)
print("Remaining:", a)
print("Removed:", a.pop())
print("Remaining:", a)
```

```python
a = ["a", "b", "c", "d", "e"]
print(a.pop(1))
print(a)
print(a.pop(3))
print(a)
print(a.pop(-2))
print(a)
print(a.pop(-1))
print(a)
```

## Choosing a list or tuple

Choose based on how the data should behave, not just which literal is shorter.

| Situation | Usual choice | Reason |
|---|---|---|
| A collection will gain, lose, or replace items | List | Supports in-place changes and dynamic size |
| A collection represents similar items processed together | Often a list | Communicates a collection-oriented role |
| Positions have stable meanings in a fixed record | Often a tuple | Communicates a fixed grouping |
| Coordinates or an RGB color | Often a tuple | Groups a fixed number of related values |
| A function returns several related results | Often a tuple | Supports packing and caller-side unpacking |

These are conventions, not rules about allowed item types. A list can be heterogeneous, and a tuple can be homogeneous.

Tuple immutability protects the tuple’s positions, not the internal state of every object it contains. It should not be treated as a guarantee that an entire nested data structure cannot change.

### Returning related values
A function can return one tuple containing multiple results. The caller can store that tuple or unpack it immediately.
